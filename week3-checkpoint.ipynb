## Using the provided simulator

The simulator is provided through the `GG4` package. The main object you will work with is:

- `brain = Brain(random_seed=...)`

You can think of the simulator as a dynamical system that evolves over time:
- you observe noisy neural activity,
- optionally apply stimulation inputs,
- and study how the system responds.

The simulator exposes only a limited interface. You do **not** have direct access to the internal state of the system.

Useful methods and attributes:

- `brain.input_dim` — required length of the command vector `u(t)`
- `brain.next_state(u)` — advance the brain by one time step using input `u`
- `brain.next_state()` — advance one step with no input
- `brain.measure()` — read the noisy observation `y(t)`
- `brain.current_time_stamp` — current time index

### Important note about inputs
The elements of `u(t)` are clipped internally to the range `[0, 1]`.
That means:
- negative values will not help,
- very large values will not have additional effect once clipped,
- it is sensible to think about **input scaling** and **control effort**.

The following code provides an example of how to interact with the simulator.

```python
#Initial state finder for estimator given random noise input
import numpy as np

def collect_system_data(brain, num_steps):
    """
    Probes the system with random inputs to collect data for System ID.
    """
    input_dim = brain.input_dim
    # Assuming we can measure once to get the output dimension
    initial_y = brain.measure()
    output_dim = len(initial_y) if isinstance(initial_y, (list, np.ndarray)) else 1
    
    # Pre-allocate arrays
    U = np.zeros((input_dim, num_steps))
    Y = np.zeros((output_dim, num_steps))
    
    print(f"Starting data harvest for {num_steps} steps...")
    
    for k in range(num_steps):
        # 1. Generate random input strictly bounded between 0.0 and 1.0
        u_k = np.random.uniform(0.0, 1.0, size=(input_dim,))
        
        # 2. Apply input to the system
        brain.next_state(u_k)
        
        # 3. Measure the resulting output
        y_k = brain.measure()
        
        # 4. Store data (as column vectors)
        U[:, k] = u_k
        Y[:, k] = y_k

    return U, Y
```

```python
import numpy as np
import scipy.linalg as la
from dataclasses import dataclass
import warnings

@dataclass
class LDSParams:
    A: np.ndarray   # (d, d)   dynamics
    B: np.ndarray   # (d, m)   input
    C: np.ndarray   # (N, d)   observation
    Q: np.ndarray   # (d, d)   process noise cov
    R: np.ndarray   # (N, N)   observation noise cov
    mu_0: np.ndarray  # (d,)   initial state mean
    P_0: np.ndarray   # (d, d) initial state cov

def validate_system_data(Y: np.ndarray, U: np.ndarray):
    """Robustness checks to ensure matrices won't crash the SVD or EM steps."""
    if Y.ndim != 2 or U.ndim != 2:
        raise ValueError(f"Data must be 2D arrays. Got Y:{Y.ndim}D, U:{U.ndim}D")
        
    if Y.shape[0] != U.shape[0]:
        raise ValueError(f"Time dimension mismatch: Y has {Y.shape[0]} steps, U has {U.shape[0]}.")
        
    if np.isnan(Y).any() or np.isnan(U).any():
        raise ValueError("Data contains NaNs. Clean data before system ID.")
        
    if np.isinf(Y).any() or np.isinf(U).any():
        raise ValueError("Data contains Infs. Clean data before system ID.")
        
    T, N = Y.shape
    M = U.shape[1]
    
    if T < (2 * N + 2 * M + 10):
        warnings.warn("Very short time series provided. Subspace ID may be rank-deficient or overfit.")
        
    return T, N, M

def build_hankel(Z: np.ndarray, num_lags: int) -> np.ndarray:
    T, dim = Z.shape
    T_h = T - num_lags + 1
    return np.vstack([Z[i:i+T_h].T for i in range(num_lags)])

# =====================================================================
# 1. THE N4SID INITIALIZATION (Required to start EM safely)
# =====================================================================
def n4sid(Y: np.ndarray, U: np.ndarray, d: int, num_lags: int) -> tuple[np.ndarray, LDSParams]:
    T, N = Y.shape
    M = U.shape[1]
    T_h = T - num_lags + 1

    H_y = build_hankel(Y, num_lags)
    H_u = build_hankel(U, num_lags)

    # Oblique Projection
    Qu, _ = la.qr(H_u.T, mode='economic')
    O_hat = H_y - (H_y @ Qu) @ Qu.T

    # SVD & Subspace Extraction
    Ug, Sg, _ = la.svd(O_hat, full_matrices=False)
    Gamma = Ug[:, :d] * np.sqrt(Sg[:d])
    C = Gamma[:N, :]

    # State extraction & Regression
    X_cols = la.pinv(Gamma) @ O_hat
    X_sub = X_cols.T

    n_reg = T_h - 1
    LHS = np.hstack([X_sub[1:], Y[:n_reg, :]]) 
    RHS = np.hstack([X_sub[:-1], U[:n_reg, :]]) 

    Theta, _, _, _ = la.lstsq(RHS, LHS)
    Theta = Theta.T

    A = Theta[:d, :d]
    B = Theta[:d, d:]
    
    # Process & Obs Noise Covariance
    X_pred = (A @ X_sub[:-1].T + B @ U[:n_reg].T).T
    Q = np.cov((X_sub[1:] - X_pred).T)
    Q = 0.5 * (Q + Q.T) + np.eye(d) * 1e-6

    R = np.cov((Y[:T_h] - (C @ X_sub.T).T).T)
    R = 0.5 * (R + R.T) + np.eye(N) * 1e-6

    # Forward Simulate
    X_hat = np.zeros((T, d))
    X_hat[:T_h] = X_sub
    for t in range(T_h, T):
        X_hat[t] = A @ X_hat[t-1] + B @ U[t-1]

    return X_hat, LDSParams(A, B, C, Q, R, X_hat[0], np.eye(d)*1.0)


# =====================================================================
# 2. THE MASTER FUNCTION (Auto-tune + EM Refinement)
# =====================================================================
def identify_system_em(
    Y: np.ndarray,
    U: np.ndarray,
    forced_latent_dim: int = None, # Set this to force dimensions
    max_latent_dim: int = 20,
    max_iter: int = 30,
    tol: float = 1e-4
) -> tuple[np.ndarray, LDSParams, dict]:
    """
    Unified function for System ID. 
    1. Validates inputs for mathematical robustness.
    2. Auto-selects latent dimensions (unless forced by the user).
    3. Generates N4SID initial guess.
    4. Refines via Expectation-Maximization (RTS Smoother).
    """
    # --- 1. Robustness Checks ---
    T, N, M = validate_system_data(Y, U)
    
    # Determine lags
    rough_lags = min(max(10, 2 * (forced_latent_dim or max_latent_dim)), T // 3)
    
    # --- 2. Dimension Selection ---
    if forced_latent_dim is not None:
        #print(f"Bypassing auto-selection. Forcing LatentDim = {forced_latent_dim}")
        best_d = forced_latent_dim
        info = {'LatentDim': best_d, 'num_lags': rough_lags, 'forced': True}
    else:
        #print("Auto-selecting LatentDim based on singular value spectrum...")
        H_y = build_hankel(Y, rough_lags)
        H_u = build_hankel(U, rough_lags)
        Qu, _ = la.qr(H_u.T, mode='economic')
        O_hat = H_y - (H_y @ Qu) @ Qu.T
        _, S, _ = la.svd(O_hat, full_matrices=False)
        
        # Select dimension explaining 95% of variance
        cum_var = np.cumsum(S**2) / np.sum(S**2)
        best_d = int(np.argmax(cum_var >= 0.95)) + 1
        best_d = min(best_d, max_latent_dim)
        print(f"  -> Selected LatentDim = {best_d} (explains {cum_var[best_d-1]:.1%} variance)")
        info = {'LatentDim': best_d, 'num_lags': rough_lags, 'forced': False, 'singular_vals': S}

    # Safety check: Ensure lags are still valid for the chosen dimension
    if rough_lags * N <= best_d:
        raise ValueError(f"num_lags ({rough_lags}) * outputs ({N}) must be > LatentDim ({best_d})")

    # --- 3. N4SID Initialization ---
    _, lds = n4sid(Y, U, best_d, rough_lags)
    A, B, C, Q, R, mu0, P0 = lds.A, lds.B, lds.C, lds.Q, lds.R, lds.mu_0, lds.P_0

    #print(f"Starting EM Refinement (Max Iterations: {max_iter})...")
    ll_prev = -np.inf

    # --- 4. Expectation-Maximization Loop ---
    for iteration in range(max_iter):
        
        # E-STEP: RTS Kalman Smoother
        mu_f, P_f = np.zeros((T, best_d)), np.zeros((T, best_d, best_d))
        mu_p, P_p = np.zeros((T, best_d)), np.zeros((T, best_d, best_d))
        ll = 0.0

        mu_p[0], P_p[0] = mu0, P0

        # Forward Pass (Filter)
        for t in range(T):
            innov = Y[t] - C @ mu_p[t]
            S_inn = C @ P_p[t] @ C.T + R
            
            # Robust inverse for Innovation Covariance
            S_inn_inv = la.inv(S_inn)
            
            sign, logdet = np.linalg.slogdet(S_inn)
            ll += -0.5 * (N * np.log(2 * np.pi) + logdet + innov @ S_inn_inv @ innov)

            K_gain = P_p[t] @ C.T @ S_inn_inv
            ImKC = np.eye(best_d) - K_gain @ C
            mu_f[t] = mu_p[t] + K_gain @ innov
            
            # Joseph form for numerical stability
            P_f[t] = ImKC @ P_p[t] @ ImKC.T + K_gain @ R @ K_gain.T

            if t < T - 1:
                mu_p[t+1] = A @ mu_f[t] + B @ U[t]
                P_p[t+1] = A @ P_f[t] @ A.T + Q

        # Backward Pass (Smoother)
        mu_s, P_s = mu_f.copy(), P_f.copy()
        P_cross = np.zeros((T - 1, best_d, best_d))

        for t in range(T - 2, -1, -1):
            # Robust inverse for smoothing
            P_p_inv = la.pinv(P_p[t + 1]) 
            G = P_f[t] @ A.T @ P_p_inv
            
            mu_s[t] = mu_f[t] + G @ (mu_s[t+1] - mu_p[t+1])
            P_s[t] = P_f[t] + G @ (P_s[t+1] - P_p[t+1]) @ G.T
            P_cross[t] = P_s[t+1] @ G.T

        # M-STEP: Parameter Updates
        S00 = np.sum([P_s[t] + np.outer(mu_s[t], mu_s[t]) for t in range(T-1)], axis=0)
        S11 = np.sum([P_s[t+1] + np.outer(mu_s[t+1], mu_s[t+1]) for t in range(T-1)], axis=0)
        S10 = np.sum([P_cross[t] + np.outer(mu_s[t+1], mu_s[t]) for t in range(T-1)], axis=0)
        Su0 = np.sum([np.outer(U[t], mu_s[t]) for t in range(T-1)], axis=0)
        Suu = np.sum([np.outer(U[t], U[t]) for t in range(T-1)], axis=0)
        Su1 = np.sum([np.outer(U[t], mu_s[t+1]) for t in range(T-1)], axis=0)

        LHS_AB = np.block([[S00, Su0.T], [Su0, Suu]])
        RHS_AB = np.hstack([S10.T, Su1.T])
        AB = la.solve(LHS_AB.T, RHS_AB.T).T
        A, B = AB[:, :best_d], AB[:, best_d:]

        # Update Q
        Q = (S11 - A@S10.T - B@Su1 - S10@A.T - Su1.T@B.T + A@S00@A.T + A@Su0.T@B.T + B@Su0@A.T + B@Suu@B.T) / (T - 1)        
        Q = 0.5 * (Q + Q.T) + np.eye(best_d) * 1e-6

        # Update C
        Sxx = np.sum([P_s[t] + np.outer(mu_s[t], mu_s[t]) for t in range(T)], axis=0)
        Syx = np.sum([np.outer(Y[t], mu_s[t]) for t in range(T)], axis=0)
        C = Syx @ la.inv(Sxx)

        # Update R
        R = np.zeros((N, N))
        for t in range(T):
            ex = mu_s[t]
            exx = P_s[t] + np.outer(ex, ex)
            R += np.outer(Y[t], Y[t]) - C@np.outer(ex, Y[t]) - np.outer(Y[t], ex)@C.T + C@exx@C.T
        R = 0.5 * (R / T + (R / T).T) + np.eye(N) * 1e-6

        mu0, P0 = mu_s[0], P_s[0]

        # Convergence Check
        delta_ll = ll - ll_prev
        #print(f"  Iter {iteration+1:02d} | Log-Likelihood: {ll:.4f} | Δ: {delta_ll:+.4e}")
        if iteration > 0 and abs(delta_ll) < tol:
            #print("  -> EM Converged.")
            break
        ll_prev = ll

    lds = LDSParams(A, B, C, Q, R, mu0, P0)
    return mu_s, lds, info
```

```python
#state estimator used instead of reacalcualting the state matrices each time
import numpy as np
import scipy.linalg as la

class AugmentedKalmanFilter:
    def __init__(self, A: np.ndarray, B: np.ndarray, C: np.ndarray, 
                 Q_process: np.ndarray, R_measure: np.ndarray, Q_disturbance: float = 0.01):
        """
        Initializes the Augmented Kalman Filter for Offset-Free Tracking.
        
        Q_disturbance: A tuning scalar. Higher values make the filter trust the 
                       disturbance estimate more (faster adaptation to steady-state error, 
                       but more sensitive to random noise).
        """
        self.d = A.shape[0]
        self.m = B.shape[1]
        self.N = C.shape[0]
        
        # 1. Build Augmented Matrices
        # A_aug = [A  0]
        #         [0  I]
        self.A_aug = np.block([
            [A, np.zeros((self.d, self.N))],
            [np.zeros((self.N, self.d)), np.eye(self.N)]
        ])
        
        # B_aug = [B]
        #         [0]
        self.B_aug = np.vstack([B, np.zeros((self.N, self.m))])
        
        # C_aug = [C  I]
        self.C_aug = np.hstack([C, np.eye(self.N)])
        
        # 2. Build Augmented Process Noise Covariance
        Q_d = np.eye(self.N) * Q_disturbance
        self.Q_aug = la.block_diag(Q_process, Q_d)
        
        # 3. Calculate Steady-State Kalman Gain (L)
        P_prior = la.solve_discrete_are(self.A_aug.T, self.C_aug.T, self.Q_aug, R_measure)
        S = self.C_aug @ P_prior @ self.C_aug.T + R_measure
        self.L_aug = P_prior @ self.C_aug.T @ la.inv(S)
        
        # 4. Initialize Internal Augmented State: [x; d]
        self.x_aug = np.zeros((self.d + self.N, 1))

    def update(self, y_k: float, u_prev: float) -> tuple[np.ndarray, np.ndarray]:
        """
        Takes the current measurement and the previous input, updates the filter, 
        and returns the separated state and disturbance estimates.
        """
        # Ensure correct shapes
        y_k = np.array(y_k).reshape(-1, 1)
        u_prev = np.array(u_prev).reshape(-1, 1)
        
        # 1. Predict current output based on previous state
        y_pred = self.C_aug @ self.x_aug
        
        # 2. Filter Step: Update estimate using the measurement error
        self.x_aug = self.x_aug + self.L_aug @ (y_k - y_pred)
        
        # 3. Extract the separated states
        x_est = self.x_aug[:self.d]
        d_est = self.x_aug[self.d:]
        
        # 4. Predict Step: Advance internal model for the *next* loop iteration
        self.x_aug = self.A_aug @ self.x_aug + self.B_aug @ u_prev
        
        return x_est, d_est
```

```python
from scipy.optimize import minimize
import numpy as np

class ConstrainedMPC:
    def __init__(self, A, B, C, horizon, Q_cost, R_cost):
        self.A = A
        self.B = B
        self.C = C
        self.N = horizon
        self.Q = Q_cost  
        self.R = R_cost  
        
        self.d = A.shape[0]
        self.m = B.shape[1]
        
        # Warm start guess
        self.u_guess = np.zeros(self.N * self.m)

    def get_input(self, x_current, target_trajectory_window):
        def cost_function(u_flat):
            u_seq = u_flat.reshape(self.N, self.m)
            cost = 0
            x = x_current.copy().reshape(-1, 1)
            
            for i in range(self.N):
                u_i = u_seq[i].reshape(-1, 1)
                x = self.A @ x + self.B @ u_i
                y_pred = self.C @ x
                
                # --- UPDATE THIS LINE ---
                # Force the target slice to be a column vector to match y_pred
                y_ref = target_trajectory_window[i].reshape(-1, 1)
                error = y_pred - y_ref
                
                cost += (error.T @ self.Q @ error)[0, 0]
                cost += (u_i.T @ self.R @ u_i)[0, 0]
                
            return cost

        # Strict bounds
        bounds = [(0.0, 1.0) for _ in range(self.N * self.m)]
        
        # RELAXED SPEED CONSTRAINTS: 
        # ftol tightened for maximum precision, maxiter removed to allow full convergence
        result = minimize(
            cost_function, 
            self.u_guess, 
            bounds=bounds, 
            method='L-BFGS-B',
            options={'ftol': 1e-6, 'disp': False} 
        )
        
        optimal_u_seq = result.x.reshape(self.N, self.m)
        self.u_guess[:-self.m] = result.x[self.m:]
        self.u_guess[-self.m:] = optimal_u_seq[-1] 
        
        return optimal_u_seq[0]
    
import numpy as np
from scipy.linalg import solve_discrete_are

class LQR:
    def __init__(self, A, B, C, Q_cost_y, R_cost):
        """
        Initializes the LQR controller and pre-computes the optimal gain matrix.
        """
        self.A = A
        self.B = B
        self.C = C
        self.n_states = A.shape[0]
        self.n_inputs = B.shape[1]
        self.n_outputs = C.shape[0]
        
        # 1. Transform the Y-space penalty into an X-space penalty
        # Since MPC cost is (Cx - y_ref)^T Q (Cx - y_ref), the state penalty is C^T Q C
        self.Q_x = self.C.T @ Q_cost_y @ self.C
        self.R = R_cost
        
        # 2. Solve the Discrete Algebraic Riccati Equation (DARE) once offline
        P = solve_discrete_are(self.A, self.B, self.Q_x, self.R)
        
        # 3. Compute the optimal feedback gain matrix K
        # K = (B^T P B + R)^-1 B^T P A
        self.K = np.linalg.inv(self.B.T @ P @ self.B + self.R) @ (self.B.T @ P @ self.A)
        
        # 4. Precompute the pseudo-inverse for Steady-State Reference Tracking
        # We need to find the ideal steady-state [x_ss, u_ss] that yields y_ref.
        # Matrix M solves: [ I - A,  -B ] [x_ss] = [   0   ]
        #                  [   C  ,   0 ] [u_ss] = [ y_ref ]
        M = np.block([
            [np.eye(self.n_states) - self.A, -self.B],
            [self.C,                         np.zeros((self.n_outputs, self.n_inputs))]
        ])
        
        # We use pinv (pseudo-inverse) because m > n (16 > 2), 
        # meaning perfect tracking of all 16 channels is impossible. 
        # pinv automatically finds the best "least-squares" compromise.
        self.M_pinv = np.linalg.pinv(M)

    def get_input(self, x_current, target_trajectory_window):
        """
        Calculates the fast LQR input. 
        Signature matches MPC, but LQR only needs the immediate next target, 
        so it ignores the rest of the window.
        """
        # LQR does not look ahead, so grab just the immediate target
        y_ref = target_trajectory_window[0].reshape(-1, 1)
        x_curr = x_current.reshape(-1, 1)

        # 1. Find the target steady-state state and input to hit y_ref
        target_vector = np.vstack([np.zeros((self.n_states, 1)), y_ref])
        ss_targets = self.M_pinv @ target_vector
        
        x_ss = ss_targets[:self.n_states]
        u_ss = ss_targets[self.n_states:]

        # 2. Apply the LQR control law: u = -K(x - x_ss) + u_ss
        u_optimal = -self.K @ (x_curr - x_ss) + u_ss
        
        # 3. Enforce the hard boundaries natively handled by MPC
        u_clipped = np.clip(u_optimal.flatten(), 0.0, 1.0)
        
        return u_clipped
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import svd, hankel
# from GG4 import Brain

# --- System Initialization ---
U_dim = brain.input_dim
dummy_y = np.array(brain.measure())
Y_dim = dummy_y.shape[0]

# ==========================================
# PHASE 1: Schroeder Phase Multisine Design
# ==========================================
print("Phase 1: Synthesizing Schroeder Multisine Input...")
N_samples = 3000  # Number of time steps
F_lines = 200     # Number of frequency lines to inject

t = np.arange(N_samples)
freqs = np.linspace(0.001, 0.49, F_lines)
phases = - (np.arange(F_lines) * (np.arange(F_lines) - 1) / F_lines) * np.pi

# Build the base signal
u_base = np.zeros(N_samples)
for f_idx in range(F_lines):
    u_base += np.cos(2 * np.pi * freqs[f_idx] * t + phases[f_idx])

# Normalize to prevent exploding states
u_base = u_base / np.max(np.abs(u_base)) * 2.0 

# Create linearly independent inputs for the 2 channels
U_data = np.zeros((U_dim, N_samples))
U_data[0, :] = u_base
U_data[1, :] = np.roll(u_base, N_samples // 2) * -1.0 # Phase shift & invert

# Run the system to collect Data
print("  Running System Data Collection...")
Y_data = np.zeros((Y_dim, N_samples))

# Settle system
for _ in range(500): brain.next_state(np.zeros(U_dim))

for k in range(N_samples):
    Y_data[:, k] = np.array(brain.measure())
    brain.next_state(U_data[:, k])


# ==========================================
# PHASE 2: OKID (Observer-Kalman Filter ID)
# ==========================================
print("Phase 2: Executing OKID to decouple dominant/hidden states...")
p = 30  # Observer order (how many past steps to look at)

# 2A. Build the Data Matrix (V)
# We map y_k to [u_k, u_{k-1}, y_{k-1}, ..., u_{k-p}, y_{k-p}]
V_rows = U_dim + p * (U_dim + Y_dim)
V_cols = N_samples - p
V_matrix = np.zeros((V_rows, V_cols))
Y_target = Y_data[:, p:]

for k in range(p, N_samples):
    col = [U_data[:, k]]
    for i in range(1, p + 1):
        col.append(U_data[:, k-i])
        col.append(Y_data[:, k-i])
    V_matrix[:, k-p] = np.concatenate(col)

# 2B. Solve Least Squares to find Observer Parameters (M)
# Using pseudoinverse for supreme numerical stability
print("  Solving Large Scale Least Squares...")
M = Y_target @ np.linalg.pinv(V_matrix)

# 2C. Recover True System Markov Parameters (Y_k = C * A^(k-1) * B)
print("  Recovering True System Markov Parameters...")
m_params = 100 # Number of true system Markov parameters to generate
Y_markov = np.zeros((m_params, Y_dim, U_dim))

# Extract D matrix (Direct Feedthrough)
Y_markov[0] = M[:, 0:U_dim] 

# Recursion variables
Y_bar_u = np.zeros((p + 1, Y_dim, U_dim))
Y_bar_y = np.zeros((p + 1, Y_dim, Y_dim))

for i in range(1, p + 1):
    idx = U_dim + (i - 1) * (U_dim + Y_dim)
    Y_bar_u[i] = M[:, idx : idx + U_dim]
    Y_bar_y[i] = M[:, idx + U_dim : idx + U_dim + Y_dim]

# The OKID Recovery Recursion Formula
for k in range(1, m_params):
    term = np.zeros((Y_dim, U_dim))
    if k <= p:
        term += Y_bar_u[k]
    
    limit = min(k, p)
    for i in range(1, limit + 1):
        term += Y_bar_y[i] @ Y_markov[k - i]
        
    Y_markov[k] = term


# ==========================================
# PHASE 3: ERA (Eigensystem Realization)
# ==========================================
print("Phase 3: Building Hankel Matrix & Performing SVD...")
# We build a block Hankel matrix from our recovered, cleaned Markov parameters
alpha = 40  # Block rows
beta = 40   # Block columns

# Initialize empty Hankel matrix
H = np.zeros((alpha * Y_dim, beta * U_dim))

for i in range(alpha):
    for j in range(beta):
        # We start from k=1 (ignoring D matrix at k=0)
        H[i*Y_dim : (i+1)*Y_dim, j*U_dim : (j+1)*U_dim] = Y_markov[i + j + 1]

# The Moment of Truth: SVD of the Hankel Matrix
U_svd, Sigma, Vh_svd = svd(H)

# ==========================================
# VISUALIZATION
# ==========================================
plt.figure(figsize=(10, 6))
# Plot only the top 50 singular values, as the rest are guaranteed computational zero
plt.semilogy(range(1, 51), Sigma[:50], marker='o', linestyle='-', color='indigo')

plt.title("OKID/ERA: Hankel Singular Values (System Order Identification)")
plt.xlabel("Singular Value Index (State Number)")
plt.ylabel("Singular Value Magnitude ($\sigma_i$)")
plt.grid(True, which='both', ls='--', alpha=0.7)

# Add a threshold line to guide the eye
plt.axhline(y=Sigma[0]*1e-4, color='r', linestyle=':', label="Expected Noise Floor")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import pinv
# from GG4 import Brain # Make sure this is imported in your environment

# ==========================================
# PHASE 1: Data Collection (Schroeder Multisine)
# ==========================================
# brain = Brain(random_seed=0)
U_dim = brain.input_dim
Y_dim = np.array(brain.measure()).shape[0]

print("Phase 1: Synthesizing Schroeder Multisine & Collecting Data...")
N_samples = 2000  # Reduced slightly to keep EM computation time manageable
F_lines = 100

t = np.arange(N_samples)
freqs = np.linspace(0.001, 0.49, F_lines)
phases = - (np.arange(F_lines) * (np.arange(F_lines) - 1) / F_lines) * np.pi

u_base = np.zeros(N_samples)
for f_idx in range(F_lines):
    u_base += np.cos(2 * np.pi * freqs[f_idx] * t + phases[f_idx])
u_base = u_base / np.max(np.abs(u_base)) * 2.0 

U_data = np.zeros((U_dim, N_samples))
U_data[0, :] = u_base
U_data[1, :] = np.roll(u_base, N_samples // 2) * -1.0

Y_data = np.zeros((Y_dim, N_samples))

# Settle system
for _ in range(200): brain.next_state(np.zeros(U_dim))

# Collect data
for k in range(N_samples):
    Y_data[:, k] = np.array(brain.measure())
    brain.next_state(U_data[:, k])


# ==========================================
# PHASE 2: Core EM Functions
# ==========================================

def generate_stable_random_matrices(p, q, n):
    """Generates purely random but strictly stable initial matrices."""
    A_raw = np.random.randn(n, n)
    max_eigenvalue = np.max(np.abs(np.linalg.eigvals(A_raw)))
    A_guess = (A_raw / max_eigenvalue) * 0.9 # Force stability
    
    B_guess = np.random.randn(n, q) * 0.1
    C_guess = np.random.randn(p, n) * 0.1
    
    Q_guess = np.eye(n) * 1e-2
    R_guess = np.eye(p) * 1e-2
    
    return A_guess, B_guess, C_guess, Q_guess, R_guess

def run_em_algorithm(Y, U, n_states, A, B, C, Q, R, iterations=30):
    """Executes the stabilized EM Algorithm."""
    p, T = Y.shape
    q = U.shape[0]
    n = n_states
    
    x_filt = np.zeros((n, T))
    P_filt = np.zeros((n, n, T))
    x_smooth = np.zeros((n, T))
    P_smooth = np.zeros((n, n, T))
    P_cross = np.zeros((n, n, T))
    
    log_likelihood_history = []

    for it in range(iterations):
        # --- E-STEP: Forward Kalman Filter ---
        x_pred = np.zeros(n)
        P_pred = np.eye(n)
        ll = 0
        
        for t in range(T):
            y_err = Y[:, t] - C @ x_pred
            # The 1e-8 stabilizer prevents matrix collapse from ill-conditioning
            S = C @ P_pred @ C.T + R + 1e-8 * np.eye(p) 
            K = P_pred @ C.T @ pinv(S)
            
            x_filt[:, t] = x_pred + K @ y_err
            P_filt[:, :, t] = (np.eye(n) - K @ C) @ P_pred
            
            sign, logdet = np.linalg.slogdet(S)
            ll -= 0.5 * (p * np.log(2 * np.pi) + logdet + y_err.T @ pinv(S) @ y_err)
            
            if t < T - 1:
                x_pred = A @ x_filt[:, t] + B @ U[:, t]
                P_pred = A @ P_filt[:, :, t] @ A.T + Q
                
        log_likelihood_history.append(ll)

        # --- E-STEP: Backward RTS Smoother ---
        x_smooth[:, -1] = x_filt[:, -1]
        P_smooth[:, :, -1] = P_filt[:, :, -1]
        
        for t in range(T - 2, -1, -1):
            P_pred_next = A @ P_filt[:, :, t] @ A.T + Q
            J = P_filt[:, :, t] @ A.T @ pinv(P_pred_next + 1e-8 * np.eye(n))
            
            x_smooth[:, t] = x_filt[:, t] + J @ (x_smooth[:, t+1] - (A @ x_filt[:, t] + B @ U[:, t]))
            P_smooth[:, :, t] = P_filt[:, :, t] + J @ (P_smooth[:, :, t+1] - P_pred_next) @ J.T
            P_cross[:, :, t+1] = J @ P_smooth[:, :, t+1]

        # --- M-STEP: Maximum Likelihood Parameter Update ---
        Phi = np.zeros((n, n))
        Phi_prev = np.zeros((n, n))
        Phi_cross = np.zeros((n, n))
        Y_X = np.zeros((p, n))
        X_U = np.zeros((n, q))
        Xprev_U = np.zeros((n, q))
        U_U = U @ U.T
        
        for t in range(T):
            Phi += P_smooth[:, :, t] + np.outer(x_smooth[:, t], x_smooth[:, t])
            Y_X += np.outer(Y[:, t], x_smooth[:, t])
            if t > 0:
                Phi_prev += P_smooth[:, :, t-1] + np.outer(x_smooth[:, t-1], x_smooth[:, t-1])
                Phi_cross += P_cross[:, :, t] + np.outer(x_smooth[:, t], x_smooth[:, t-1])
                X_U += np.outer(x_smooth[:, t], U[:, t-1])
                Xprev_U += np.outer(x_smooth[:, t-1], U[:, t-1])
        
        C = Y_X @ pinv(Phi)
        R = np.zeros((p, p))
        for t in range(T):
            err = Y[:, t] - C @ x_smooth[:, t]
            R += np.outer(err, err) + C @ P_smooth[:, :, t] @ C.T
        R = R / T
        
        Left_block = np.hstack((Phi_cross, X_U))
        Right_block = np.vstack((np.hstack((Phi_prev, Xprev_U)), np.hstack((Xprev_U.T, U_U))))
        
        AB_update = Left_block @ pinv(Right_block)
        A = AB_update[:, 0:n]
        B = AB_update[:, n:n+q]
        Q = (Phi - A @ Phi_cross.T - Phi_cross @ A.T + A @ Phi_prev @ A.T) / (T - 1)

    # Compute BIC
    num_params = n**2 + n*q + p*n + (n*(n+1))/2 + (p*(p+1))/2
    final_ll = log_likelihood_history[-1]
    bic = -2 * final_ll + num_params * np.log(T)
    
    return A, B, C, Q, R, bic, log_likelihood_history


# ==========================================
# PHASE 3: Multi-Start Execution Pipeline
# ==========================================
print("\nPhase 3: Running Multi-Start EM Order Identification...")
max_candidate_order = 20  # Adjust based on expected complexity and CPU time
starts_per_order = 10   # Number of random seeds per dimension
em_iterations = 50       # EM steps per seed

best_bics = []
optimal_matrices = {}

for candidate_n in range(1, max_candidate_order + 1):
    print(f"  Evaluating state dimension n = {candidate_n}...")
    
    best_ll_for_n = -np.inf
    best_bic_for_n = np.inf
    best_matrices_for_n = None
    
    for seed in range(starts_per_order):
        # Initialize with a stable random state
        A_init, B_init, C_init, Q_init, R_init = generate_stable_random_matrices(Y_dim, U_dim, candidate_n)
        
        # Run EM
        try:
            A, B, C, Q, R, bic, ll_hist = run_em_algorithm(
                Y_data, U_data, candidate_n, 
                A_init, B_init, C_init, Q_init, R_init, 
                iterations=em_iterations
            )
            final_ll = ll_hist[-1]
            
            # Keep the best performing seed for this dimension
            if final_ll > best_ll_for_n:
                best_ll_for_n = final_ll
                best_bic_for_n = bic
                best_matrices_for_n = (A, B, C, Q, R)
                
        except np.linalg.LinAlgError:
            print(f"    Seed {seed+1} encountered numerical instability. Skipping.")
            continue
            
    best_bics.append(best_bic_for_n)
    optimal_matrices[candidate_n] = best_matrices_for_n
    print(f"    -> Best BIC for n={candidate_n}: {best_bic_for_n:.2f}")

# ==========================================
# PHASE 4: Analysis & Visualization
# ==========================================
optimal_n = np.argmin(best_bics) + 1
print(f"\n✅ True System Order Identified: n = {optimal_n}")

plt.figure(figsize=(10, 6))
plt.plot(range(1, max_candidate_order + 1), best_bics, marker='o', linewidth=2, color='teal')
plt.title("Bayesian Information Criterion (BIC) vs. State Dimension", fontsize=14)
plt.xlabel("Candidate State Dimension (n)", fontsize=12)
plt.ylabel("BIC Score (Lower is Better)", fontsize=12)
plt.axvline(x=optimal_n, color='red', linestyle='--', label=f'Optimal Order: n = {optimal_n}')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()

# To extract the final matrices for the optimal system:
# A_opt, B_opt, C_opt, Q_opt, R_opt = optimal_matrices[optimal_n]
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_residual_diagnostics(Y_val, U_val, lds, channel_to_plot=0):
    """
    Plots the absolute residuals of Open-Loop vs 1-Step-Ahead prediction 
    to diagnose VAF drop-off (Drift vs. SNR Illusion).
    """
    T, N = Y_val.shape
    d = lds.A.shape[0]
    
    # ==========================================
    # 1. Calculate Open-Loop Predictions
    # ==========================================
    x_ol = lds.mu_0.copy()
    Y_pred_ol = np.zeros_like(Y_val)
    
    for t in range(T):
        Y_pred_ol[t] = lds.C @ x_ol
        x_ol = lds.A @ x_ol + lds.B @ U_val[t]
        
    err_ol = np.abs(Y_val[:, channel_to_plot] - Y_pred_ol[:, channel_to_plot])
    
    # ==========================================
    # 2. Calculate 1-Step-Ahead Predictions
    # ==========================================
    x_1s = lds.mu_0.copy()
    P_1s = lds.P_0.copy()
    Y_pred_1s = np.zeros_like(Y_val)
    
    for t in range(T):
        Y_pred_1s[t] = lds.C @ x_1s
        
        # Kalman Update
        innov = Y_val[t] - Y_pred_1s[t]
        S = lds.C @ P_1s @ lds.C.T + lds.R
        try:
            K = P_1s @ lds.C.T @ np.linalg.pinv(S)
        except np.linalg.LinAlgError:
            K = np.zeros((d, N))
            
        x_filt = x_1s + K @ innov
        P_filt = (np.eye(d) - K @ lds.C) @ P_1s
        
        # Predict Next
        x_1s = lds.A @ x_filt + lds.B @ U_val[t]
        P_1s = lds.A @ P_filt @ lds.A.T + lds.Q
        
    err_1s = np.abs(Y_val[:, channel_to_plot] - Y_pred_1s[:, channel_to_plot])
    
    # ==========================================
    # 3. Plotting
    # ==========================================
    true_signal = np.abs(Y_val[:, channel_to_plot])
    time_axis = np.arange(T)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
    
    # Plot Open-Loop
    ax1.plot(time_axis, true_signal, label="True Signal Magnitude", color='lightgray', alpha=0.8)
    ax1.plot(time_axis, err_ol, label="Absolute Error (Open-Loop)", color='red', alpha=0.8)
    ax1.set_title("Open-Loop Prediction (Diagnosing Drift)")
    ax1.set_ylabel("Amplitude")
    ax1.legend()
    ax1.grid(True, linestyle='--', alpha=0.5)
    
    # Plot 1-Step-Ahead
    ax2.plot(time_axis, true_signal, label="True Signal Magnitude", color='lightgray', alpha=0.8)
    ax2.plot(time_axis, err_1s, label="Absolute Error (1-Step-Ahead)", color='teal', alpha=0.8)
    ax2.set_title("1-Step-Ahead Prediction (Diagnosing SNR Illusion)")
    ax2.set_xlabel("Time Steps")
    ax2.set_ylabel("Amplitude")
    ax2.legend()
    ax2.grid(True, linestyle='--', alpha=0.5)
    
    plt.tight_layout()
    plt.show()

# --- How to call it at the bottom of your script ---
plot_residual_diagnostics(res_dominant.Y_val, res_dominant.U_val, res_dominant.lds, channel_to_plot=0)
```
