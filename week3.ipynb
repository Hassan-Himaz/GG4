# Week 3: Control (Individual)

In Week 3, you will build on your estimation approach to design a strategy for influencing the system.

## Objective

Your goal is to design a control policy that uses the available information (observations and/or your estimator) to generate inputs \(u(t)\) that produce a desired behaviour.

There is **no single correct solution**. You are expected to propose, justify, implement, and evaluate your own approach.

## What does “control” mean here?

Depending on your design choices, your controller may aim to:

- stabilise part of the system dynamics,
- drive the system toward a desired state,
- track a target trajectory,
- regulate or suppress certain behaviours,
- shape the system response over time.

Your task is not simply to produce outputs, but to design a strategy that meaningfully interacts with the system dynamics.

## Tasks

You should:

1. **Define a control objective**  
   Clearly specify the behaviour you want to achieve and explain why it is meaningful.

2. **Design a control strategy**  
   Propose one or more methods for generating inputs based on observations and/or estimated latent representations. Where possible, compare different approaches and justify your final design choices.

3. **Implement your controller**  
   Integrate your control policy with the simulation framework developed in Week 1 and your estimation pipeline from Week 2.

4. **Test in closed loop**  
   Run experiments where your controller continuously interacts with the system over time.

5. **Evaluate performance**  
   Use plots and quantitative measures to analyse:
   - stability,
   - responsiveness,
   - control effort,
   - robustness to noise,
   - sensitivity to parameter choices.

6. **Iterate and refine**  
   Improve your approach based on observed behaviour and failed attempts.

## Guidance

- You are encouraged to reuse and adapt your estimation approach from Week 2.
- Simple, well-justified methods are preferable to complex but poorly understood ones.

```python
import sys
import platform
import subprocess
import pathlib

ROOT = pathlib.Path(".")
WHEEL_ROOT = ROOT / "wheels"

system = platform.system()
machine = platform.machine().lower()

# -------- 1. Check Python Version --------
supported_versions = {(3, 11), (3, 12), (3, 13)}
py_ver = sys.version_info[:2]

if py_ver not in supported_versions:
    raise RuntimeError(
        f"Unsupported Python version: {py_ver[0]}.{py_ver[1]}.\n"
        "This notebook only supports Python 3.11, 3.12, or 3.13."
    )

py_tag = f"cp{py_ver[0]}{py_ver[1]}"

# -------- 2. Find Pattern Beased on Python Version and Platform --------
patterns = []

if system == "Windows":
    if machine in ("amd64", "x86_64"):
        patterns = [
            f"*{py_tag}*win_amd64.whl",
        ]
    elif machine in ("x86", "i386", "i686"):
        patterns = [
            f"*{py_tag}*win32.whl",
        ]
    else:
        raise RuntimeError(
            f"Unsupported Windows architecture: {machine}.\n"
            "Expected amd64/x86_64 or x86/i386/i686."
        )

elif system == "Darwin":
    # Prioritize universal2, then specific architectures
    if machine in ("arm64", "aarch64"):
        patterns = [
            f"*{py_tag}*macosx*universal2.whl",
            f"*{py_tag}*macosx*arm64.whl",
        ]
    elif machine in ("x86_64", "amd64"):
        patterns = [
            f"*{py_tag}*macosx*universal2.whl",
            f"*{py_tag}*macosx*x86_64.whl",
        ]
    else:
        raise RuntimeError(
            f"Unsupported macOS architecture: {machine}.\n"
            "Expected x86_64/amd64 or arm64/aarch64."
        )

elif system == "Linux":
    # Prioritize manylinux, then musllinux, for each architecture
    if machine in ("x86_64", "amd64"):
        patterns = [
            f"*{py_tag}*manylinux*x86_64.whl",
            f"*{py_tag}*musllinux*x86_64.whl",
        ]
    elif machine in ("i686", "x86"):
        patterns = [
            f"*{py_tag}*manylinux*i686.whl",
            f"*{py_tag}*musllinux*i686.whl",
        ]
    elif machine in ("arm64", "aarch64"):
        patterns = [
            f"*{py_tag}*manylinux*aarch64.whl",
            f"*{py_tag}*musllinux*aarch64.whl",
        ]
    else:
        raise RuntimeError(
            f"Unsupported Linux architecture: {machine}."
        )

else:
    raise RuntimeError(
        f"Unsupported operating system: {system}."
    )

# -------- 3. Find wheel --------
wheel = None
for pattern in patterns:
    matches = sorted(WHEEL_ROOT.rglob(pattern))
    if matches:
        wheel = matches[0]
        break

if wheel is None:
    raise RuntimeError(
        f"No compatible wheel found for:\n"
        f"  Python tag: {py_tag}\n"
        f"  System: {system}\n"
        f"  Architecture: {machine}\n\n"
        f"Searched under: {WHEEL_ROOT}\n"
        f"Patterns tried:\n  " + "\n  ".join(patterns)
    )

# -------- 4. Install --------
print(f"Installing wheel: {wheel.name}")
subprocess.check_call([
    sys.executable, "-m", "pip", "install", "--quiet", str(wheel)
])
print("Package installed.")
```

## Using the provided simulator

The simulator is provided through the `GG4` package. The main object you will work with is:

- `brain = Brain(random_seed=...)`

You can think of the simulator as a dynamical system that evolves over time:
- you observe noisy neural activity,
- optionally apply stimulation inputs,
- and study how the system responds.

The simulator exposes only a limited interface. You do **not** have direct access to the internal state of the system.

Useful methods and attributes:

- `brain.input_dim` — required length of the command vector `u(t)`
- `brain.next_state(u)` — advance the brain by one time step using input `u`
- `brain.next_state()` — advance one step with no input
- `brain.measure()` — read the noisy observation `y(t)`
- `brain.current_time_stamp` — current time index

### Important note about inputs
The elements of `u(t)` are clipped internally to the range `[0, 1]`.
That means:
- negative values will not help,
- very large values will not have additional effect once clipped,
- it is sensible to think about **input scaling** and **control effort**.

The following code provides an example of how to interact with the simulator.

```python
import numpy as np
import matplotlib.pyplot as plt
from GG4 import Brain

# Create a new brain instance
brain = Brain(random_seed=0)

# Input dimension of the brain (storyline: the number of electrodes implanted in the brain for exciting)
U = brain.input_dim
print("Brain input dimension =", U)

# There are three available interfaces to interact with the brain:
## 1. 'brain.current_time_stamp': int, 
#   the current timepoint in the experiment. It can only be the rvalue (has no setter).
print("\nMethod 1: brain.current_time_stamp")
print("current_time_stamp =", brain.current_time_stamp)  # should be 0 at the beginning of the experiment

## 2. 'brain.measure': Callable[void, List[float]], 
#   measure the current neural activity. The output is a 1D array of shape (Neurons,).
print("\nMethod 2: brain.measure()")
measurement = np.array(brain.measure())  # convert list to numpy array for better analysis
print("shape of measurement =", measurement.shape)  # should be (Neurons,)

## 3. 'brain.next_state': Callable[Optional[none|List[float]], void], 
#   apply the input to the brain and move to the next timepoint. The input should be a 1D array of shape (U,).
#   If the input is None, it means no stimulation is applied to the brain at this timepoint.
print("\nMethod 3: brain.next_state(input)")
print("current time point: ", brain.current_time_stamp)  # should be 0
brain.next_state()  # empty input, move to the next timepoint
print("current time point: ", brain.current_time_stamp)  # should be 1
input_to_apply = np.random.rand(U)  # random input for testing
brain.next_state(input_to_apply)  # apply input and move to the next timepoint
print("current time point: ", brain.current_time_stamp)  # should be 2
```

```python
import sys
if 'controller' in sys.modules:
    del sys.modules['controller']
import controller as ctrl

import numpy as np
import matplotlib.pyplot as plt
from GG4 import Brain

print("Using controller from:", ctrl.__file__)

INPUT_DIM = 2
LATENT_DIM = 5
TRIM = 15
R_WEIGHT = 0.3
Q_WEIGHT = 1.0
INPUT_SMOOTHING = 0.25
FEEDBACK_CLIP = 0.20
MIN_ACTIVATION = 0.05   # 保证两个通道都激活 -> 不会只走单 mode

# ===== 整个 notebook 只开这一个连续 brain =====
brain = Brain(random_seed=0)

# 1. mixed 激励辨识（不是白噪声）
Y_id, U_id = ctrl.collect_continuous_mixed(brain, n_steps=1000, input_dim=INPUT_DIM, seed=0)
params = ctrl.identify(Y_id, U_id, LATENT_DIM)
rho = np.max(np.abs(np.linalg.eigvals(params["A"])))
print("Identification done. spectral radius =", round(rho, 4))
print("mean |y| =", round(np.mean(np.linalg.norm(Y_id, axis=1)), 2))
print("per-channel mean input =", np.round(U_id.mean(0), 3))

# 2. LQR
K = ctrl.design_lqr(params["A"], params["B"], q_weight=Q_WEIGHT, r_weight=R_WEIGHT)
print("LQR K shape:", K.shape)

# 3. 连续经验标定
levels = np.linspace(0, 1, 31)
records = ctrl.calibrate_continuous_references(
    brain=brain, params=params, input_dim=INPUT_DIM,
    levels=levels, hold=120, tail=50, settle=30)
lo_norm = records[0]["norm_ss"]; hi_norm = records[-1]["norm_ss"]
print("reachable |y| range:", round(lo_norm, 2), "to", round(hi_norm, 2))

plt.figure(figsize=(8, 4.5))
plt.plot([r["a"] for r in records], [r["norm_ss"] for r in records], "o-")
plt.xlabel("input level a"); plt.ylabel("empirical steady |y|")
plt.title("Empirical calibration table"); plt.tight_layout(); plt.show()

# target 取在可达范围内
T_LOW  = round(lo_norm + 0.30 * (hi_norm - lo_norm), 1)
T_HIGH = round(lo_norm + 0.70 * (hi_norm - lo_norm), 1)
T_MID  = round(0.5 * (T_LOW + T_HIGH), 1)
print("targets: LOW =", T_LOW, " MID =", T_MID, " HIGH =", T_HIGH)
```

```python
import sys
if 'controller' in sys.modules:
    del sys.modules['controller']
import controller as ctrl
import numpy as np
import matplotlib.pyplot as plt

def washout_to_level(brain, level=0.25, steps=100):
    """非重置 washout：把连续 brain 推到温和状态（真实输入序列，不是 reset）"""
    u = np.ones(INPUT_DIM) * level
    for _ in range(steps):
        brain.next_state(u)

def plot_tracking(y_log, u_log, target_seq, title, trim=TRIM, smooth=9, step=False):
    yn = np.linalg.norm(y_log, axis=1)
    target_seq = np.asarray(target_seq, float)
    tt = np.arange(len(target_seq)); keep = tt >= trim
    plt.figure(figsize=(11, 4.5))
    if step:
        plt.step(tt[keep], target_seq[keep], where="post", color="k", ls="--", label="target")
    else:
        plt.plot(tt[keep], target_seq[keep], "k--", label="target")
    plt.plot(tt[keep], yn[keep], alpha=0.25, label="actual raw")
    plt.plot(tt[keep], ctrl.moving_average(yn, smooth)[keep], lw=2.5, label="actual smoothed")
    plt.xlabel("time"); plt.ylabel("|y|"); plt.title(title); plt.legend()
    plt.tight_layout(); plt.show()
    plt.figure(figsize=(11, 2.6))
    plt.plot(tt[keep], u_log[keep, 0], label="u0 (electrode 0)")
    plt.plot(tt[keep], u_log[keep, 1], label="u1 (electrode 1)")
    plt.ylim(-0.05, 1.05); plt.xlabel("time"); plt.ylabel("input u")
    plt.title(title + " — inputs"); plt.legend(); plt.tight_layout(); plt.show()
```

```python
import sys
if 'controller' in sys.modules:
    del sys.modules['controller']
import controller as ctrl
import numpy as np, matplotlib.pyplot as plt

washout_to_level(brain, level=0.25, steps=100)

N = 200
setpoint = np.full(N, T_HIGH)
y_on, u_on, ref_on, reach_on, a_on = ctrl.run_empirical_lqr(
    brain, params, K, setpoint, records,
    input_smoothing=INPUT_SMOOTHING, feedback_clip=FEEDBACK_CLIP,
    reference_mode="interp", min_activation=MIN_ACTIVATION)
m_on = ctrl.evaluate_run(y_on, u_on, target_norm=T_HIGH, trim=TRIM)
usage = ctrl.channel_usage(u_on)

# 同一连续 brain 上的无控制基线
y_off = ctrl.run_open_loop_constant(brain, np.zeros(INPUT_DIM), N)
off_norm = np.linalg.norm(y_off, axis=1)

print("controlled steady |y| =", round(m_on["steady_norm"], 2), " (target", T_HIGH, ")")
print("uncontrolled |y| mean =", round(off_norm.mean(), 2))
print("chatter =", round(m_on["chatter"], 3), " saturation =", f"{m_on['saturation']:.1%}")
print("channel balance =", round(usage["balance"], 3), "(1.0=完全均衡)")
print("per-channel mean input =", np.round(usage["per_channel_mean"], 3))

on_norm = np.linalg.norm(y_on, axis=1)
tt = np.arange(N); keep = tt >= TRIM
fig, ax = plt.subplots(2, 1, figsize=(11, 6.5))
ax[0].axhline(T_HIGH, color="k", ls="--", label="target")
ax[0].plot(tt[keep], on_norm[keep], color="tab:blue", alpha=0.25, label="controlled (raw)")
ax[0].plot(tt[keep], ctrl.moving_average(on_norm, 10)[keep], color="tab:blue", lw=2.5, label="controlled (smoothed)")
ax[0].plot(tt[keep], off_norm[keep], color="tab:red", alpha=0.6, label="uncontrolled")
ax[0].set_ylabel("|y|"); ax[0].set_title("Setpoint: controlled vs uncontrolled"); ax[0].legend()
ax[1].plot(tt[keep], u_on[keep, 0], label="u0 (electrode 0)")
ax[1].plot(tt[keep], u_on[keep, 1], label="u1 (electrode 1)")
ax[1].axhline(0, color="gray", lw=.5); ax[1].axhline(1, color="gray", lw=.5)
ax[1].set_ylim(-0.05, 1.05); ax[1].set_ylabel("input u"); ax[1].set_xlabel("time")
ax[1].set_title("Control inputs (both channels active)"); ax[1].legend()
fig.tight_layout(); plt.show()
```

```python
import sys
if 'controller' in sys.modules:
    del sys.modules['controller']
import controller as ctrl
import numpy as np, matplotlib.pyplot as plt

def measure_response(brain, u, settle=35, hold=100, tail=40):
    u = np.asarray(u, float)
    for _ in range(settle):
        brain.next_state(np.clip(u, 0, 1))
    Ys = []
    for _ in range(hold):
        Ys.append(np.array(brain.measure(), float))
        brain.next_state(np.clip(u, 0, 1))
    Ys = np.array(Ys)
    tn = np.linalg.norm(Ys[-tail:], axis=1)
    return float(np.mean(tn)), float(np.std(tn))

lv = np.linspace(0, 1, 11)
sweeps = {
    "u0 only": [np.array([a, 0.0]) for a in lv],
    "u1 only": [np.array([0.0, a]) for a in lv],
    "u0=u1":   [np.array([a, a])   for a in lv],
}
res = {}
for name, inputs in sweeps.items():
    means, stds = [], []
    for u in inputs:
        mu, sd = measure_response(brain, u)
        means.append(mu); stds.append(sd)
    res[name] = (np.array(means), np.array(stds))
    print(f"{name}: |y| from {means[0]:.1f} to {means[-1]:.1f}")

plt.figure(figsize=(9, 5))
for name, (mu, sd) in res.items():
    plt.errorbar(lv, mu, yerr=sd, marker="o", capsize=3, label=name)
plt.xlabel("input level"); plt.ylabel("steady-state |y|")
plt.title("Input-channel effect and saturation"); plt.legend(); plt.tight_layout(); plt.show()

plt.figure(figsize=(9, 4))
for name, (mu, sd) in res.items():
    gain = np.diff(mu) / np.diff(lv); mid = 0.5 * (lv[1:] + lv[:-1])
    plt.plot(mid, gain, "o-", label=name)
    if np.max(gain) > 1e-9:
        idx = np.where(gain < 0.30 * np.max(gain))[0]
        if len(idx): print(f"{name}: saturation begins near level {mid[idx[0]]:.2f}")
        else: print(f"{name}: no clear saturation in [0,1]")
plt.xlabel("input level midpoint"); plt.ylabel("marginal gain d|y|/du")
plt.title("Saturation check"); plt.legend(); plt.tight_layout(); plt.show()
```

```python
# ============================================================
# Big-gap figure 1:
# target miss of the OLD model-inversion method
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

def model_reference_from_norm(params, target_norm, levels=np.linspace(0, 1, 201)):
    """
    Old method:
    choose u=[a,a] using model-predicted steady-state norm only.
    """
    A, B, C = params["A"], params["B"], params["C"]
    d = params.get("d", np.zeros(A.shape[0]))
    ym = params["y_mean"]
    I = np.eye(A.shape[0])

    best = None
    for a in levels:
        u = np.ones(B.shape[1]) * float(a)
        try:
            x_ss = np.linalg.solve(I - A, B @ u + d)
        except np.linalg.LinAlgError:
            x_ss = np.linalg.pinv(I - A) @ (B @ u + d)

        y_ss = C @ x_ss + ym
        pred_norm = np.linalg.norm(y_ss)
        err = abs(pred_norm - target_norm)

        if best is None or err < best["err"]:
            best = {
                "a": float(a),
                "u": u.copy(),
                "pred_norm": float(pred_norm),
                "err": float(err),
            }
    return best["a"], best["u"], best["pred_norm"]


def measure_real_steady_norm(brain, u, settle=60, hold=120, tail=50):
    """
    On the continuous brain, apply constant u and measure real steady |y|.
    """
    u = np.asarray(u, float)

    for _ in range(settle):
        brain.next_state(np.clip(u, 0, 1))

    Ys = []
    for _ in range(hold):
        y = np.array(brain.measure(), float)
        Ys.append(y)
        brain.next_state(np.clip(u, 0, 1))

    Ys = np.array(Ys)
    return float(np.mean(np.linalg.norm(Ys[-tail:], axis=1)))


def washout_to_level(brain, level=0.25, steps=100):
    u = np.ones(INPUT_DIM) * level
    for _ in range(steps):
        brain.next_state(u)


targets = np.array([13.0, 15.0, 17.0, 18.0, 19.0])

old_a = []
old_model_reachable = []
old_real_achieved = []

for tg in targets:
    washout_to_level(brain, level=0.25, steps=100)

    a, u_old, pred = model_reference_from_norm(params, tg)
    real = measure_real_steady_norm(brain, u_old, settle=50, hold=100, tail=40)

    old_a.append(a)
    old_model_reachable.append(pred)
    old_real_achieved.append(real)

old_a = np.array(old_a)
old_model_reachable = np.array(old_model_reachable)
old_real_achieved = np.array(old_real_achieved)

print(f"{'target':>8}{'old a':>10}{'model says':>14}{'real achieved':>16}{'miss':>12}")
for tg, a, pm, rr in zip(targets, old_a, old_model_reachable, old_real_achieved):
    print(f"{tg:>8.2f}{a:>10.3f}{pm:>14.2f}{rr:>16.2f}{(rr-tg):>12.2f}")

plt.figure(figsize=(9, 5))
plt.plot(targets, targets, 'k--', lw=2, label='desired target')
plt.plot(targets, old_model_reachable, 'o-', lw=2, label='old method: model-predicted reachable')
plt.plot(targets, old_real_achieved, 'o-', lw=2, label='old method: real achieved')
plt.xlabel('desired target |y|')
plt.ylabel('steady-state |y|')
plt.title('Why pure model inversion is misleading')
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
plt.bar([str(int(t)) for t in targets], old_real_achieved - targets)
plt.axhline(0, color='k', ls='--')
plt.xlabel('desired target |y|')
plt.ylabel('real achieved - target')
plt.title('Target miss when using old model inversion')
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 实验: Week3 LQR 跟踪振荡参考 vs 直接开环注入振荡
# 问题: LQR(为稳态设计)能否产生 BMI band-pass 需要的振荡神经活动?
# 前提: 已跑过 Week3 的辨识 cell, 有 brain, params, K, records
# ============================================================
import numpy as np
import matplotlib.pyplot as plt

PERIOD = 16          # BMI 选肩肌肉的周期
N = 300
t = np.arange(N)

# 振荡参考: 让 ‖y‖ 在可达范围内按周期16正弦
amp = 0.3 * (hi_norm - lo_norm)
center = 0.5 * (lo_norm + hi_norm)
osc_ref = center + amp * np.sin(2*np.pi*t / PERIOD)

# ---- 方法A: 用 Week3 LQR 跟踪这个振荡参考 ----
washout_to_level(brain, level=0.25, steps=80)
yA, uA, refA, _, _ = ctrl.run_empirical_lqr(
    brain, params, K, osc_ref, records,
    input_smoothing=0.05,        # 弱平滑, 让它能快速变化
    feedback_clip=0.5,           # 松clip, 允许大修正
    reference_mode="interp", min_activation=0.05)
ynA = np.linalg.norm(yA, axis=1)

# ---- 方法B: 直接开环注入正弦 input (就是Week4做的) ----
washout_to_level(brain, level=0.25, steps=80)
yB = []
for k in range(N):
    u_val = 0.5 + 0.5*np.sin(2*np.pi*k / PERIOD)
    yB.append(np.array(brain.measure(), float))
    brain.next_state(np.clip(np.full(2, u_val), 0, 1))
yB = np.array(yB)
ynB = np.linalg.norm(yB, axis=1)

# ---- 分析: 两种方法下, 神经活动通道0 的振荡频率和幅度 ----
def dominant(sig):
    s = sig[50:] - sig[50:].mean()
    sp = np.abs(np.fft.rfft(s)); f = np.fft.rfftfreq(len(s))
    pk = f[1+np.argmax(sp[1:])]
    return (1/pk if pk>0 else np.inf), s.std()

pa, aa = dominant(yA[:,0])
pb, ab = dominant(yB[:,0])
print(f"目标周期 = {PERIOD}")
print(f"方法A (LQR跟踪): 神经活动主周期≈{pa:.1f}, 振幅={aa:.3f}, 输入chatter={np.mean(np.abs(np.diff(uA,axis=0))):.3f}")
print(f"方法B (开环注入): 神经活动主周期≈{pb:.1f}, 振幅={ab:.3f}")

# ---- 画图对比 ----
fig, ax = plt.subplots(3, 1, figsize=(11, 9))
keep = t >= 30
ax[0].plot(t[keep], osc_ref[keep], 'k--', label='oscillating reference ‖y‖')
ax[0].plot(t[keep], ynA[keep], 'tab:blue', alpha=0.6, label='LQR tracking ‖y‖')
ax[0].plot(t[keep], ynB[keep], 'tab:orange', alpha=0.6, label='open-loop inject ‖y‖')
ax[0].set_ylabel("‖y‖"); ax[0].legend(); ax[0].set_title(f"Track vs inject (period={PERIOD})")

ax[1].plot(t[keep], yA[keep,0], 'tab:blue', label='LQR: neural ch0')
ax[1].plot(t[keep], yB[keep,0], 'tab:orange', alpha=0.7, label='inject: neural ch0')
ax[1].set_ylabel("neural activity ch0"); ax[1].legend()
ax[1].set_title("Is the neural activity actually oscillating at the target frequency?")

ax[2].plot(t[keep], uA[keep,0], label='LQR input u0')
ax[2].plot(t[keep], uA[keep,1], label='LQR input u1')
ax[2].set_ylabel("LQR input"); ax[2].set_xlabel("time"); ax[2].legend()
ax[2].set_title("LQR's input — does it oscillate cleanly?")
plt.tight_layout(); plt.show()
```
